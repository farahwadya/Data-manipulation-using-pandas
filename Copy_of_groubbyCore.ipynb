```python
import pandas as pd
import numpy as np
from statistics import mode
```

```python
inspath='/content/drive/MyDrive/Axsos Data science/AXSOSACADEMY/01-Fundamentals/Week02/Data/insurance_mod.csv'
```

```python
df= pd.read_csv(inspath)
```

## average age

```python
avg_mean= df['age'].mean()
print (f'average age = {round(avg_mean,2)}')
```

### max age

```python
max_age= df['age'].max()
print(f'oldest age = {max_age}')
```

### min age

```python
min_age= df['age'].min()
print(f'oldest age = {min_age}')
```

### average for charges

```python
charge_avg=df['charges'].mean()
print(f'average of charges = {round(charge_avg,3)}')
```

 ### average for charges grouped by sex

```python
charge_avg_sex = df.groupby('sex')['charges'].mean()
print(f'average of charges groubed by sex= {round(charge_avg_sex,3)}')
```

### average for charges, grouped by if the person is a smoker

```python
charge_avg_smoker= df.groupby(df['smoker']== True)['charges'].mean()
print(f'average for charges of the smoker {round(charge_avg_smoker,3)}')
```

### What is the average for charges grouped by if the person smokes and their sex?

```python
chareg_avg_smoke_sex=df.groupby(['smoker','sex'])['charges'].mean()
print(f'average for charges of the smoker {round(chareg_avg_smoke_sex,3)}')
```
